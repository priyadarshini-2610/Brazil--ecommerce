14. **How long does it take to move between steps:**

**purchase → approval**

**approval → shipping**

**shipping → delivery**

```python
merged_df['order_purchase_timestamp'] = pd.to_datetime(merged_df['order_purchase_timestamp'], errors='coerce')
merged_df['order_approved_at'] = pd.to_datetime(merged_df['order_approved_at'], errors='coerce')
merged_df['order_delivered_carrier_date'] = pd.to_datetime(merged_df['order_delivered_carrier_date'], errors='coerce')
merged_df['order_delivered_customer_date'] = pd.to_datetime(merged_df['order_delivered_customer_date'], errors='coerce')
```

```python
merged_df['purchase_to_approval_days'] = (merged_df['order_approved_at'] - merged_df['order_purchase_timestamp']).dt.total_seconds() / (24*3600)

merged_df['approval_to_shipping_days'] = (merged_df['order_delivered_carrier_date'] - merged_df['order_approved_at']).dt.total_seconds() / (24*3600)

merged_df['shipping_to_delivery_days'] = (merged_df['order_delivered_customer_date'] - merged_df['order_delivered_carrier_date']).dt.total_seconds() / (24*3600)
```

```python
time_summary = pd.DataFrame({
    'Step': ['Purchase - Approval', 'Approval - Shipping', 'Shipping - Delivery'],
    'Average_Days': [
        merged_df['purchase_to_approval_days'].mean(),
        merged_df['approval_to_shipping_days'].mean(),
        merged_df['shipping_to_delivery_days'].mean()
    ],
    'Median_Days': [
        merged_df['purchase_to_approval_days'].median(),
        merged_df['approval_to_shipping_days'].median(),
        merged_df['shipping_to_delivery_days'].median()
    ]
})

display(time_summary.round(2))
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Step', y='Average_Days', data=time_summary, palette='Blues_d')
plt.ylabel('Average Days')
plt.title('Average Time Between Order Steps')
```

**INSIGHT:**

*   Fast approval: The system automatically or quickly approves most orders.
*   Shipping delay is moderate: Sellers take ~1–3 days to ship.
    There are some outliers, so a few sellers may be slow.
*   Delivery is the main bottleneck: The largest contributor to total order
    time is shipping to customer — influenced by geography, remote locations, or heavier products
